#IF ELSE ELIF

```python
idade = int(input("Digite sua idade:"))
if idade < 16:
  print("Menor de idade")
elif idade >= 16 or idade < 18:
  print("Voto opcional")
elif idade >= 18:
  print("Voto obrigatório")
else:
  print("Nenhuma das opções")
```

```python
nota = float(input("Digite sua nota"))
if nota >= 7:
  print("Aprovado")
elif nota >= 5 and nota < 7:
  print("Recuperação")
elif nota < 5:
  print("Reprovado")
else:
  print("Opção inválida")
```
